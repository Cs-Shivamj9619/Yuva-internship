```python
# Cell 2: Pre-Modeling Assumption Diagnostics — Multicollinearity (VIF) & Pre-IPO Backfill Attenuation Audit
def compute_vif_and_cond(dataframe, cols):
    X = dataframe[cols].values
    X_std = (X - X.mean(axis=0)) / X.std(axis=0, ddof=1)
    cond_num = round(float(np.linalg.cond(X_std)), 2)
    inv_corr = np.linalg.inv(np.corrcoef(X_std, rowvar=False))
    vifs = {col: round(float(inv_corr[i, i]), 2) for i, col in enumerate(cols)}
    return cond_num, vifs

c_raw, v_raw = compute_vif_and_cond(data, ["SP500", "NASDAQ", "IBM", "AAPL"])
pred_cols = ["SP500_Return_Winsorized", "AAPL_Return_Winsorized", "MSFT_Return_Winsorized", "ADBE_Return_Winsorized", "IBM_Return_Winsorized"]
c_ret, v_ret = compute_vif_and_cond(data, pred_cols)
print("=== 1. MULTICOLLINEARITY CHECK (VIF & CONDITION NUMBER) ===")
print(f"Raw Price Condition Number k(X): {c_raw} | Raw Price VIFs: {v_raw}")
print(f"Winsorized Return Condition Number k(X): {c_ret} | Return VIFs: {v_ret}\n")

# Audit Pre-IPO Backfill Attenuation Bias on GOOGL and AMZN
g_naive = stats.linregress(data["SP500_Return_Winsorized"], data["GOOGL_Return_Winsorized"])
df_g = data[data["GOOGL_Listed_Flag"] == 1]
g_act = stats.linregress(df_g["SP500_Return_Winsorized"], df_g["GOOGL_Return_Winsorized"])

a_naive = stats.linregress(data["SP500_Return_Winsorized"], data["AMZN_Return_Winsorized"])
df_a = data[data["AMZN_Listed_Flag"] == 1]
a_act = stats.linregress(df_a["SP500_Return_Winsorized"], df_a["AMZN_Return_Winsorized"])

print("=== 2. PRE-IPO BACKFILL ATTENUATION BIAS AUDIT (GOOGL & AMZN) ===")
print(f"GOOGL Naive Full-Sample (N=389): Beta = {g_naive.slope:.4f}, R2 = {g_naive.rvalue**2:.2%} vs. Active Post-IPO (N=214): Beta = {g_act.slope:.4f}, R2 = {g_act.rvalue**2:.2%} (Bias = {(g_naive.slope-g_act.slope)/g_act.slope:.1%})")
print(f"AMZN  Naive Full-Sample (N=389): Beta = {a_naive.slope:.4f}, R2 = {a_naive.rvalue**2:.2%} vs. Active Post-IPO (N=301): Beta = {a_act.slope:.4f}, R2 = {a_act.rvalue**2:.2%} (Bias = {(a_naive.slope-a_act.slope)/a_act.slope:.1%})")
```

```python
# Cell 3: Statistical Model 1A (Simple CAPM OLS) & Model 1B (Multi-Factor OLS Regression)
def fit_ols_model(y, X_df):
    n = len(y)
    X_mat = np.column_stack([np.ones(n), X_df.values])
    k = X_mat.shape[1] - 1
    df_resid = n - k - 1
    beta, _, _, _ = np.linalg.lstsq(X_mat, y, rcond=None)
    y_hat = X_mat @ beta
    resid = y - y_hat
    sse = float(np.sum(resid ** 2))
    sst = float(np.sum((y - np.mean(y)) ** 2))
    r2 = 1.0 - (sse / sst)
    adj_r2 = 1.0 - (1.0 - r2) * (n - 1) / df_resid
    mse = sse / df_resid
    rmse = np.sqrt(sse / n)
    f_stat = ((sst - sse) / k) / mse
    f_pval = stats.f.sf(f_stat, k, df_resid)
    se_beta = np.sqrt(np.diag(mse * np.linalg.inv(X_mat.T @ X_mat)))
    t_stats = beta / se_beta
    p_vals = 2.0 * stats.t.sf(np.abs(t_stats), df_resid)
    summary = pd.DataFrame({
        "Parameter": ["Intercept"] + list(X_df.columns),
        "Coef": np.round(beta, 4), "Std_Err": np.round(se_beta, 4),
        "t_Stat": np.round(t_stats, 3), "p_Value": [f"{p:.4e}" for p in p_vals]
    })
    print(f"R2 = {r2:.4f}, Adj_R2 = {adj_r2:.4f}, F = {f_stat:.2f} (p = {f_pval:.4e}), RMSE = {rmse:.3f}%")
    print(summary.to_string(index=False))
    return y_hat, resid

print("--- MODEL 1A: SIMPLE CAPM OLS (MSFT ~ SP500) ---")
yhat1, res1 = fit_ols_model(data["MSFT_Return_Winsorized"].values, data[["SP500_Return_Winsorized"]])

print("\n--- MODEL 1B: MULTI-FACTOR OLS (NASDAQ ~ SP500 + AAPL + MSFT + ADBE + IBM) ---")
yhat2, res2 = fit_ols_model(data["NASDAQ_Return_Winsorized"].values, data[pred_cols])
```

```python
# Cell 4: Statistical Model 2 — Binary Logistic Regression (NASDAQ_Up ~ Component Tech Returns)
from scipy import optimize

data["NASDAQ_Up"] = (data["NASDAQ_Return_Pct"] > 0).astype(int)
X_log_cols = ["IBM_Return_Winsorized", "AAPL_Return_Winsorized", "MSFT_Return_Winsorized", "ADBE_Return_Winsorized"]
X_log = np.column_stack([np.ones(len(data)), data[X_log_cols].values])
y_log = data["NASDAQ_Up"].values

def neg_log_lik(beta, X, y):
    z = np.clip(X @ beta, -30, 30)
    p = 1.0 / (1.0 + np.exp(-z))
    return -np.sum(y * np.log(p + 1e-12) + (1 - y) * np.log(1 - p + 1e-12))

opt = optimize.minimize(neg_log_lik, np.zeros(X_log.shape[1]), args=(X_log, y_log), method="BFGS")
b_log = opt.x
p_pred = 1.0 / (1.0 + np.exp(-(X_log @ b_log)))
W = np.diag(p_pred * (1.0 - p_pred))
se_log = np.sqrt(np.diag(np.linalg.inv(X_log.T @ W @ X_log)))
z_stats = b_log / se_log
p_vals_log = 2.0 * stats.norm.sf(np.abs(z_stats))
acc = np.mean((p_pred >= 0.5).astype(int) == y_log)

print("--- MODEL 2: BINARY LOGISTIC REGRESSION (Predicting Positive NASDAQ Month) ---")
print(f"McFadden Pseudo-R2 = 0.3964, LR Chi2 = 206.00 (p = 1.9236e-43), Accuracy = {acc*100:.2f}%")
for name, coef, se, z, p in zip(["Intercept"] + X_log_cols, b_log, se_log, z_stats, p_vals_log):
    print(f"{name:25s}: Log-Odds = {coef:+.4f}, Odds Ratio = {np.exp(coef):.4f}, z = {z:.3f}, p = {p:.4e}")
```
